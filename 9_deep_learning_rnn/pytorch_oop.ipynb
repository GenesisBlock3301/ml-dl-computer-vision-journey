```python
import torch
import torch.nn as nn

layer = nn.Linear(10, 5)
activation = nn.ReLU()

# create a random input tensor: batch_size=3, input_features=10
input_tensor = torch.randn(3, 10)

# forward pass
output = activation(layer(input_tensor))
print(output.shape)  # torch.Size([3, 5])
```
