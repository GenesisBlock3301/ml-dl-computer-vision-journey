```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

torch.manual_seed(1)
# https://medium.com/@smrati.katiyar/explaining-embedding-layer-in-pytorch-1f22b88c1a69
```

```python
embedding_layer = nn.Embedding(num_embeddings=10, embedding_dim=3)
print("embedding_layer:", embedding_layer)

input_indices = torch.tensor([1,2,3,4])
print("input_indices:", input_indices)

output = embedding_layer(input_indices)
print(output)
```
