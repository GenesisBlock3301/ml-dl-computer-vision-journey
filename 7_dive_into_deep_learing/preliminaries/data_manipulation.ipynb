```python
import torch

x = torch.arange(12, dtype=torch.float32)
x
```

```python
# Number of elements in x
x.numel()
```

```python
# Shape of x
x.shape
```

```python
# Reshape of x
X = x.reshape((3, 4))
X
```

```python
x.reshape(3, -1)
```

# Indexing & Slicing

```python
X[-1], X[1:3]
```

```python
X[1, 2] = 17
X
```

```python
X[:2, 0:1] = 12
X
```

# Operation

```python
# Exponential
torch.exp(x)
```

```python
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])

x + y, x - y, x*y, x / y, x ** y
```

```python
# Concatenation
X = torch.arange(12, dtype=torch.float32).reshape((3, 4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
torch.cat((X, Y), dim=0), torch.cat((X, Y), dim=1)
```

```python
X == Y
```

# Broadcasting

```python
a = torch.arange(3).reshape((3, 1))
b = torch.arange(2).reshape((1, 2))
a, b
```

```python
 a + b
```

# Saving Memory

```python
before = id(Y)
Y = Y + X
id(Y) == before
```

```python
Z = torch.zeros_like(Y)
print("id(Z)", id(Z))
Z[:] = X + Y
print("id(Z)", id(Z))
```
